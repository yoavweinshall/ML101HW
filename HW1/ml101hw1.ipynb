```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
data = np.concatenate([X_train, X_test], axis=0).reshape(-1, 28 * 28)
labels = np.concatenate([y_train, y_test], axis=0)
```

```python
import numpy.random
idx = numpy.random.RandomState(0).choice(70000, 11000)
train = data[idx[:10000], :].astype(int)
train_labels = labels[idx[:10000]]
test = data[idx[10000:], :].astype(int)
test_labels = labels[idx[10000:]]
```

### a. 

```python
def kNN(train: np.ndarray, train_labels: np.ndarray, query_image: np.ndarray, k: int) -> int:
    """
    A simple implementation of the k-Nearest Neighbors algorithm.
    Args:
        train: A 2D numpy array of shape (num_train_samples, num_features) representing the training data.
        train_labels: A 1D numpy array of shape (num_train_samples,) representing the labels for the training data.
        query_image: A 1D numpy array of shape (num_features,) representing the query image.
        k: An integer representing the number of nearest neighbors to consider.
    Returns:
        An integer representing the predicted label for the query image.
    """
    train = np.asarray(train)
    train_labels = np.asarray(train_labels)
    query_image = np.asarray(query_image).reshape(-1)

    if len(train) != len(train_labels):
        raise ValueError("train and train_labels must have the same number of samples")
    if not 1 <= k <= len(train):
        raise ValueError("k must be between 1 and the number of training samples")
    if query_image.shape[0] != train.shape[1]:
        raise ValueError("query_image must have the same number of features as train")

    distances = np.sum((train - query_image) ** 2, axis=1)
    nearest_indices = np.argsort(distances, kind="stable")[:k]
    nearest_labels = train_labels[nearest_indices]
    labels, counts = np.unique(nearest_labels, return_counts=True)
    return int(labels[np.argmax(counts)])
```

### b. 

```python
def return_right_predictions(train_size: int, k: int) -> int:
    """
    Returns the number of correct predictions made by the kNN algorithm on the test set.
    Args:
        train_size: An integer representing the number of training samples to use.
        k: An integer representing the number of nearest neighbors to consider.
    Returns:
        An integer representing the number of correct predictions made by the kNN algorithm on the test set.
    """
    right_answers = 0
    for i in range(len(test)):
        prediction = kNN(train[:train_size].reshape(-1, 28 * 28), train_labels[:train_size], test[i].reshape(-1, 28 * 28), k)
        if prediction == test_labels[i]:
            right_answers += 1
    return right_answers

print(f"Accuracy: {return_right_predictions(1000, 10) / len(test) * 100}%")
```

### C

```python
right_answers=np.array([])
for k in range(1,101):
    right_answers=np.append(right_answers, return_right_predictions(1000, k))

plt.plot(range(1, 101), right_answers / len(test) * 100)
plt.xlabel('k')
plt.ylabel('correct predictions (%)')
plt.title('kNN Accuracy vs. k')
plt.grid()
plt.show()

```

### d

```python
right_answers=np.array([])
for n in range(100, 5001, 100):
    right_answers=np.append(right_answers, return_right_predictions(n, k=1))

plt.plot(range(1, 5001, 100), right_answers / len(test) * 100)
plt.xlabel('Training Size')
plt.ylabel('correct predictions (%)')
plt.title('kNN Accuracy vs. Training Size')
plt.grid()
plt.show()
```
